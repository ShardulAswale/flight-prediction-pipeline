# 04 - Visualisation
Plot trajectories on an interactive map.

```python
%load_ext autoreload
%autoreload 2

import os, sys
sys.path.append(os.path.abspath('..'))
import pandas as pd
from src.utils import load_config
from src.visualisation import TrajectoryVisualiser
```

```python
config = load_config('../configs/config.yaml')
traj_path = f"../{config['paths']['processed_data_dir']}/{config['paths']['trajectories_file']}"
if os.path.exists(traj_path):
    df_traj = pd.read_parquet(traj_path)
    print(f"Loaded {len(df_traj['trajectory_id'].unique())} unique trajectories.")
else:
    print('No trajectories found.')
    df_traj = pd.DataFrame()
```

```python
# Display a single random trajectory map
import random

if not df_traj.empty:
    traj_ids = list(df_traj['trajectory_id'].unique())
    if len(traj_ids) > 0:
        random_traj = random.choice(traj_ids)
        print(f'Displaying random trajectory: {random_traj}')
        m = TrajectoryVisualiser.generate_interactive_map(df_traj, trajectory_ids=[random_traj])
        display(m)
    else:
        print('No valid trajectories to display.')
else:
    print('No data available to display.')
```
